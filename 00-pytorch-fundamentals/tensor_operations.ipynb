Manipulating tensors

- `+ - * /` with a single number apply the operation to every element
- The result is a new tensor; the original is unchanged
- Store it back to keep the change: `x = x + 10`
- Division always gives float results, even for whole-number tensors
- Function versions: `torch.add`, `torch.sub`, `torch.mul`, `torch.div`
- Tensor with tensor works position by position (same shape needed)

Block 1: What is matrix multiplication?

- Element-wise (`*`): multiply matching positions and keep them separate
- Matrix multiplication: multiply matching pairs, then add them all up
- For two vectors this is called the dot product and gives a single number
- Write it as `torch.matmul(a, b)` or `a @ b`
- A neuron's weighted sum (inputs x weights, added up) is a dot product
- Matrix multiplication is the core operation of deep learning
- `torch.matmul` is much faster than a Python for loop

 Matrix x matrix

- Each result value = one row of the first matrix dot one column of the second
- Result at [row i, column j] = row i of A dot column j of B
- (a x b) @ (b x c) gives an (a x c) result
- Element-wise `*` and matrix multiplication `@` give different results
- Matrix multiplication usually gives a different result if you swap the order

```python

t1 = torch.tensor([[1, 2], [3, 4]])  # create the first 2x2 matrix
t2 = torch.tensor([[5, 6], [7, 8]])  # create the second 2x2 matrix
print(t1.shape)                      # shape of the first matrix
print(t2.shape)                      # shape of the second matrix

print(t1 @ t2)                       # rows of t1 dot columns of t2
print(t2 @ t1)                       # rows of t2 dot columns of t1 (different result)
print(t1 * t2)                       # element-wise: matching positions only, no adding up

# Q4: add a third customer
quantity = torch.tensor([[2, 1, 3],
                         [1, 0, 2],
                         [0, 2, 1]])  # quantities: 3 customers x 3 items
prices = torch.tensor([[10, 12],
                       [20, 18],
                       [30, 25]])     # prices: 3 items x 2 shops
bill = quantity @ prices             # each customer's total bill at each shop
print(bill)                          # display the bills
print(bill.shape)                    # size of the result: 3 customers x 2 shops

```

The 2 shape rules

- Rule 1: the inner dimensions must match: (a x b) @ (b x c)
- Rule 2: the result has the outer dimensions: (a x c)
- A mismatch gives: `RuntimeError: mat1 and mat2 shapes cannot be multiplied`
- Always check the two shapes side by side before using `@`
- Shape mismatches are the most common error in deep learning

Transpose

- Transpose swaps rows and columns: a (2 x 3) becomes (3 x 2)
- Write it as `x.T` or `torch.transpose(x, 0, 1)`
- Use it to fix shape errors: (2 x 3) @ (2 x 3) fails, (2 x 3) @ (3 x 2) works
- `.T` does not change the original tensor
- Transposing twice gives back the original
- Neural network layers compute `inputs @ weights.T + bias`

Min, max, sum and mean

- `x.min()`, `x.max()`, `x.sum()`, `x.mean()` reduce a tensor to one number
- Function versions: `torch.min(x)`, `torch.max(x)`, `torch.sum(x)`, `torch.mean(x)`
- `mean` needs a float tensor; on int64 it fails with a dtype error
- Fix: convert first with `x.type(torch.float32).mean()`
- The results are one-number tensors; `.item()` gives a plain number

A week's temperatures: temps = torch.tensor([31, 29, 35, 33, 30, 28, 34])

Q1. Find the hottest day's temperature, the coldest, and the total of all seven.

Q2. Find the average temperature. First try temps.mean() directly and see what happens, then fix it.

Q3. Do Q1 and Q2 again using the function versions (torch.max, torch.min, torch.sum, torch.mean).

Q4. Print the average as a plain Python number using .item().

```python
temps = torch.tensor([31, 29, 35, 33, 30, 28, 34])  # create a tensor of a week's temperatures

hottest_day = temps.max()                  # highest temperature of the week
print(hottest_day)                         # display the highest temperature
coldest_day = torch.min(temps)             # lowest temperature, using the function version
print(coldest_day)                         # display the lowest temperature
print(temps.sum())                         # total of all seven temperatures

# print(temps.mean())                      # fails: mean needs a float tensor
print(temps.type(torch.float32).mean())    # convert to float, then find the average
print(temps.type(torch.float32).mean().item())  # average as a plain Python number
```

Other names for the same thing

- `torch.mm(A, B)` = matrix multiplication for 2-D matrices only; errors on vectors
- `torch.matmul(A, B)` and `A @ B` work on vectors and matrices
- `torch.multiply(x, y)` = `torch.mul(x, y)` = `x * y` (element-wise)
- `torch.float` = float32, `torch.half` = float16, `torch.double` = float64
- `torch.long` = int64 (error messages often say "Long")

### Catch-up Block 2: Integer datatypes

- Whole-number types: `int8`, `int16`, `int32`, `int64` (default, also called `torch.long`)
- More bits = bigger range and more memory
- `int8` holds -128 to 127; `uint8` holds 0 to 255 (common for pixel values)
- Numbers too big for the type wrap around silently, with no error
- Smaller types save memory (8-bit models are used on mobile devices)

### Catch-up Block 3: Timing code with %%time

- `%%time` on the first line of a notebook cell measures how long the cell takes
- `%time` (single %) times a single line
- "Wall time" is the real time you waited
- Magic commands work in notebooks (Jupyter, VS Code, Colab), not in .py files
- Python for loops over tensors are very slow; PyTorch operations like `@` are much faster

Positional min and max

- `x.argmax()` gives the position (index) of the largest value
- `x.argmin()` gives the position (index) of the smallest value
- Function versions: `torch.argmax(x)`, `torch.argmin(x)`
- Use the position to look up the value: `x[x.argmax()]`
- If the max appears more than once, the first position is returned
- Works on whole-number tensors (no float needed, unlike `mean`)
- Used in classification to turn scores into a predicted label

A week's temperatures, Monday to Sunday: temps = torch.tensor([31, 29, 35, 33, 30, 28, 34])

Q1. Find the position of the hottest day and the coldest day. Which days of the week are they?

Q2. Use the position from Q1 to get the hottest temperature back, like marks[marks.argmax()].

Q3. A model gives the scores torch.tensor([0.1, 0.6, 0.3]) for the labels ["cat", "dog", "bird"]. Use argmax to print the predicted label. (Hint: .item() turns the position into a plain number you can use with a Python list.)

Q4. Guess what torch.tensor([5, 9, 9, 2]).argmax() gives, then run it.

```python
temps = torch.tensor([31, 29, 35, 33, 30, 28, 34])  # a week's temperatures, Monday to Sunday
hot_day = temps.max()                    # the highest temperature
cold_day = temps.min()                   # the lowest temperature

index_hot_day = torch.argmax(temps)      # position of the hottest day
index_cold_day = torch.argmin(temps)     # position of the coldest day
print(index_hot_day, index_cold_day)     # display both positions
print(temps[index_hot_day])              # use the position to look up the hottest temperature

scores = torch.tensor([0.1, 0.6, 0.3])   # model scores for each category
labels = ["cat", "dog", "bird"]          # the category names, in the same order
print(labels[scores.argmax().item()])    # turn the highest score's position into a label

print(torch.tensor([5, 9, 9, 2]).argmax())  # max appears twice, so the first position is returned
```

```python
scores = torch.tensor([0.05, 0.15, 0.70, 0.10])  # the model's score for each fruit
labels = ["apple", "banana", "mango", "orange"]   # the fruit names, in the same order

max_score = torch.argmax(scores)
print(max_score)

python_number = max_score.item()
print(python_number)

print(labels[python_number])
```

reshape

- `x.reshape(new_shape)` rearranges the same numbers into a new shape
- The product of the new shape must equal the total number of items
- Numbers are filled in order, row by row
- A wrong size gives: `RuntimeError: shape '[5, 3]' is invalid for input of size 12`
- `reshape(1, n)` adds an extra dimension to a vector of n numbers
- `-1` in a shape means "work this size out automatically", e.g. `x.reshape(3, -1)`
- `x.reshape(-1)` flattens any tensor into one long row
- Only one `-1` is allowed, and the total must still divide evenly
- The original tensor keeps its shape; store the result to keep the new one

Start with: nums = torch.arange(1, 25) (that's 24 numbers)

Q1. Reshape nums into 4×6, 6×4 and 2×3×4. Predict each shape first, then check.

Q2. Which of these shapes work for 24 numbers: (5, 5), (8, 3), (2, 2, 6), (7, 3)? Predict, then try one that fails and read the error.

Q3. Use -1: reshape nums into (4, -1) and (-1, 3). Predict what number -1 becomes in each case, then check the shapes.

Q4. Take your 2×3×4 tensor from Q1 and flatten it back into one long row using -1. What's its shape?

Q5. Reshape torch.arange(10) into shape (1, 10). How many dimensions does it have now?

Q6. Try nums.reshape(-1, -1). What error do you get, and why?

Q7. After all the reshapes, print nums.shape. Did the original change?

```python
nums = torch.arange(1, 25)      # create 24 numbers: 1 to 24

# Q1: reshape into different shapes (store each one, then check it)
grid_4x6 = nums.reshape(4, 6)   # 4 rows of 6
print(grid_4x6.shape)           # shape of the 4x6 version
grid_6x4 = nums.reshape(6, 4)   # 6 rows of 4
print(grid_6x4.shape)           # shape of the 6x4 version
cube = nums.reshape(2, 3, 4)    # 2 trays, each with 3 rows of 4
print(cube.ndim, cube.shape)    # dimensions and shape of the 3-D version

# Q2: 24 items: (8, 3) and (2, 2, 6) work; (5, 5) and (7, 3) don't
print(nums.reshape(8, 3).shape)     # 8 x 3 = 24, so it works
print(nums.reshape(2, 2, 6).shape)  # 2 x 2 x 6 = 24, so it works
# nums.reshape(5, 5)                # fails: 5 x 5 = 25, not 24

# Q3: let PyTorch work out the -1
print(nums.reshape(4, -1).shape)    # 4 rows; columns = 24 / 4
print(nums.reshape(-1, 3).shape)    # 3 columns; rows = 24 / 3

# Q4: flatten the 3-D tensor back into one row
flat = cube.reshape(-1)             # flatten into a single row
print(flat.shape)                   # shape of the flattened tensor

# Q5: add an extra dimension to a vector
row = torch.arange(10).reshape(1, 10)  # put 10 numbers inside 1 row
print(row.ndim, row.shape)          # dimensions and shape

# Q6: two unknowns are not allowed
# nums.reshape(-1, -1)              # fails: only one -1 can be worked out

# Q7: the original is unchanged
print(nums.shape)                   # shape of the original tensor
```
