Manipulating tensors

- `+ - * /` with a single number apply the operation to every element
- The result is a new tensor; the original is unchanged
- Store it back to keep the change: `x = x + 10`
- Division always gives float results, even for whole-number tensors
- Function versions: `torch.add`, `torch.sub`, `torch.mul`, `torch.div`
- Tensor with tensor works position by position (same shape needed)

Block 1: What is matrix multiplication?

- Element-wise (`*`): multiply matching positions and keep them separate
- Matrix multiplication: multiply matching pairs, then add them all up
- For two vectors this is called the dot product and gives a single number
- Write it as `torch.matmul(a, b)` or `a @ b`
- A neuron's weighted sum (inputs x weights, added up) is a dot product
- Matrix multiplication is the core operation of deep learning
- `torch.matmul` is much faster than a Python for loop

 Matrix x matrix

- Each result value = one row of the first matrix dot one column of the second
- Result at [row i, column j] = row i of A dot column j of B
- (a x b) @ (b x c) gives an (a x c) result
- Element-wise `*` and matrix multiplication `@` give different results
- Matrix multiplication usually gives a different result if you swap the order

The 2 shape rules

- Rule 1: the inner dimensions must match: (a x b) @ (b x c)
- Rule 2: the result has the outer dimensions: (a x c)
- A mismatch gives: `RuntimeError: mat1 and mat2 shapes cannot be multiplied`
- Always check the two shapes side by side before using `@`
- Shape mismatches are the most common error in deep learning

Transpose

- Transpose swaps rows and columns: a (2 x 3) becomes (3 x 2)
- Write it as `x.T` or `torch.transpose(x, 0, 1)`
- Use it to fix shape errors: (2 x 3) @ (2 x 3) fails, (2 x 3) @ (3 x 2) works
- `.T` does not change the original tensor
- Transposing twice gives back the original
- Neural network layers compute `inputs @ weights.T + bias`

Min, max, sum and mean

- `x.min()`, `x.max()`, `x.sum()`, `x.mean()` reduce a tensor to one number
- Function versions: `torch.min(x)`, `torch.max(x)`, `torch.sum(x)`, `torch.mean(x)`
- `mean` needs a float tensor; on int64 it fails with a dtype error
- Fix: convert first with `x.type(torch.float32).mean()`
- The results are one-number tensors; `.item()` gives a plain number

Positional min and max

- `x.argmax()` gives the position (index) of the largest value
- `x.argmin()` gives the position (index) of the smallest value
- Function versions: `torch.argmax(x)`, `torch.argmin(x)`
- Use the position to look up the value: `x[x.argmax()]`
- If the max appears more than once, the first position is returned
- Works on whole-number tensors (no float needed, unlike `mean`)
- Used in classification to turn scores into a predicted label

reshape

- `x.reshape(new_shape)` rearranges the same numbers into a new shape
- The product of the new shape must equal the total number of items
- Numbers are filled in order, row by row
- A wrong size gives: `RuntimeError: shape '[5, 3]' is invalid for input of size 12`
- `reshape(1, n)` adds an extra dimension to a vector of n numbers
- `-1` in a shape means "work this size out automatically", e.g. `x.reshape(3, -1)`
- `x.reshape(-1)` flattens any tensor into one long row
- Only one `-1` is allowed, and the total must still divide evenly
- The original tensor keeps its shape; store the result to keep the new one

### Extra: dim in sum, mean, max and argmax

- Without `dim`, `sum` / `mean` / `max` / `argmax` work on the whole tensor

view

- `x.view(new_shape)` shows the same data in a new shape without copying
- A view is another name for the same data: change one and the other changes too
- `x.clone()` makes an independent copy (like NumPy's `.copy()`)
- Views are fast and save memory because nothing is copied
- Same shape rules as reshape: the total must match, and `-1` is allowed

**view vs reshape**
- `view` always shares data; if it can't, it gives an error
- `reshape` shares data when it can; otherwise it quietly makes a copy
- Use `reshape` by default; use `view` when you want a guaranteed shared-data result
- You will often see `view` in older code and tutorials

**Why view sometimes fails**
- A tensor's data is stored in memory as one long line; the shape is just a reading rule
- A view only changes the reading rule; it never moves the data
- After a transpose, the data is no longer in reading order (not contiguous)
- `view` fails on such tensors; `reshape` works by copying into a new order
- `x.is_contiguous()` checks whether the data is stored in reading order

permute

- `x.permute(...)` reorders the dimensions of a tensor
- The numbers are the old positions, written in the new order you want
- `[height, width, colour]` to `[colour, height, width]` is `permute(2, 0, 1)`
- Each value stays with its correct row, column and channel
- `reshape` to the same shape scrambles the data; use `permute` to reorder dimensions
- For 2 dimensions, `permute(1, 0)` is the same as `.T`
- `permute` shares data with the original (like `view`)

Q1. Create a random image with shape (224, 224, 3) (height, width, colour). Permute it to colour-first. Predict the numbers inside permute(...) and the new shape.

Q2. Create a random tensor with shape (10, 20, 30). Predict the shape after permute(1, 2, 0), then check.

Q3. For the matrix [[1, 2, 3], [4, 5, 6]], compare permute(1, 0) with reshape(3, 2). Same shape, but are the values the same?

Q4. Start with a colour-first image of shape (3, 28, 28) and permute it back to (28, 28, 3). Which numbers go inside permute(...)?

```python
# Q1: height-width-colour to colour-first
image = torch.rand(224, 224, 3)         # a random image: height, width, colour channels
colour_first = image.permute(2, 0, 1)   # new order: colour, height, width
print(image.shape, colour_first.shape)  # [224, 224, 3] and [3, 224, 224]

# Q2: reorder a 3-D tensor
t = torch.rand(10, 20, 30)              # a random tensor with shape [10, 20, 30]
print(t.permute(1, 2, 0).shape)         # new order: old 1, old 2, old 0

# Q3: permute vs reshape
m = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])           # 2 rows, 3 columns
print(m.permute(1, 0))                  # swap dimensions: each column stays together
print(m.reshape(3, 2))                  # pour numbers in order: columns get mixed

# Q4: colour-first back to height-width-colour
img = torch.rand(3, 28, 28)             # a colour-first image
print(img.permute(1, 2, 0).shape)       # new order: height, width, colour

```

Picking a row or a single value

- `x[i]` selects row i; `x[i, j]` selects the value at row i, column j
- `x[i][j]` gives the same result as `x[i, j]`; the comma style is preferred
- Counting starts at 0
- Negative indexes count from the end: `x[-1]` is the last row
- Each index removes one dimension: `[3, 3]` to `[3]` to a single number
- `.item()` turns a single selected value into a plain Python number

The colon

- `:` means "all" of that dimension
- `x[:, j]` selects column j (every row); `x[i, :]` selects row i (same as `x[i]`)
- `start:end` selects a range; the end is not included
- `1:` means from 1 to the end; `:2` means from the start up to 2
- A selected column comes out as a 1-D tensor (printed as a row)
- Negative indexes also work with `:`, e.g. `x[:, -1]` is the last column

Q5. Find the total sales on the 3rd day (index 2) across all shops. (Hint: select the column, then .sum().)

```python
# rows = 4 shops, columns = 3 days
sales = torch.tensor([[10, 20, 30],
                      [15, 25, 35],
                      [20, 10, 40],
                      [ 5, 50, 45]])

print(sales[:, 1])         # all rows, column 1: every shop's sales on the 2nd day
print(sales[:, -1])        # all rows, last column: every shop's sales on the last day
print(sales[3, :])         # row index 3, all columns: same as sales[3], shape [3]
print(sales[3:])           # a range from index 3 to the end: keeps the dimension, shape [1, 3]
print(sales[:2, :])        # rows 0 and 1, all columns: the first two shops
print(sales[:, 2].sum())   # column 2 (3rd day) for every shop, then add them up

```

Indexing 3-D tensors

- A 3-D tensor is indexed as `x[tray, row, column]`
- A single number picks one item and removes that dimension
- `:` keeps the whole dimension
- `x[0, 0, 0]` gives a single value; `x[0, 0]` gives a row; `x[0]` gives a whole tray
- `x[:, :, 1]` gives column 1 from every row of every tray
- Extra brackets in a result mean `:` kept a dimension of size 1
