```python
torch.manual_seed(0)        # set the seed to 0 BEFORE creating the random tensors

t1 = torch.rand(7, 7)       # create a 7x7 random tensor (same numbers every run)
t2 = torch.rand(1, 7)       # create a 1x7 random tensor (continues reading list 0)

result = t1 @ t2.T          # (7, 7) @ (7, 1): transpose t2 so the inner numbers match
print(result)               # display the result
print(result.shape)         # torch.Size([7, 1])
```

**Question:** `torch.manual_seed()` sets the random seed for the CPU. Is there a GPU equivalent? If there is, set the GPU random seed to 1234.


```python
# Yes. The CPU and the GPU each have their own random numbers, so each has its own seed function. `torch.cuda.manual_seed()` sets the seed for the GPU.
torch.cuda.manual_seed(1234)    # set the seed for the GPU's random numbers to 1234
print(torch.cuda.initial_seed()) # check the GPU seed that is currently set
```

6. Create two random tensors of shape (2, 3) and send them both to the GPU. Set `torch.manual_seed(1234)` when creating the tensors.

```python
torch.manual_seed(1234)
a = torch.rand(2,3)
torch.manual_seed(1234)
b = torch.rand(2,3)

a = a.to(device)          # move a to the GPU and store it back
b = b.to(device)          # move b to the GPU and store it back

print(a)                  # display a (should show device='cuda:0')
print(b)                  # display b (should show device='cuda:0')
```

7. Do a matrix multiplication on a and b from Exercise 6

```python
multi = a @ b.T
print(multi)
print(multi.shape)
```

8.Find the maximum and minimum values of the output from Exercise 7 (multi).

```python
max_value = multi.max()       # the highest value in the result
min_value = multi.min()       # the lowest value in the result
print(max_value, min_value)   # display both
```

9. Find the positions (indexes) of the maximum and minimum values in multi.

```python
max_position = torch.argmax(multi)
min_position = torch.argmin(multi)
print(max_position , min_position)
```
