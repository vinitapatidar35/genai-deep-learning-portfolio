 `torch.arange(start, end, step)` creates a 1-D tensor of evenly spaced numbers
- **start**: the first number (default is 0)
- **end**: where to stop; this number is **not included**
- **step**: the jump between numbers (default is 1)
- Giving only one number sets `end`, and counting starts from 0
- To include the last number, set `end` just above it
- The result is always a vector (1-D)
- Whole-number inputs give whole numbers; decimal inputs give decimals
- `torch.range` is outdated; always use `torch.arange`

- `torch.zeros_like(x)` creates a tensor of zeros with the same shape as `x`
- `torch.ones_like(x)` creates a tensor of ones with the same shape as `x`
- The shape is read from the existing tensor, so there's no need to type it
- They also copy the number type (whole numbers or decimals) of the original
- Useful when two tensors must have the same shape, e.g. for `*`

Q1. Create a random tensor with shape (2, 4). Use zeros_like to make a zeros tensor of the same shape. Print both shapes to check that they match.

Q2. Create a tiny random image of size (2, 2, 3). Use ones_like to make a matching ones tensor. Multiply the two together. Is the result the same as the original image?

Q3. Create torch.arange(1, 6), then use ones_like on it. Also create torch.ones(5). Print both. What's different about how they look?



```python
# Q1: zeros and ones with the same shape as a random tensor
rand_tensor = torch.rand(2, 4)             # create a 2x4 tensor of random numbers
zero_same = torch.zeros_like(rand_tensor)  # create zeros with the same shape as rand_tensor
one_same = torch.ones_like(rand_tensor)    # create ones with the same shape as rand_tensor
print(rand_tensor.shape)                   # size of the original tensor
print(zero_same.shape)                     # size of the zeros copy, to confirm it matches

# Q2: multiply a tiny image by a matching ones tensor
tiny_image = torch.rand(2, 2, 3)           # create a random 2x2 colour image
one_like = torch.ones_like(tiny_image)     # create ones with the same shape as the image
print(tiny_image)                          # display the original image
print(tiny_image * one_like)               # multiply each number by 1, leaving the image unchanged

# Q3: compare ones_like with torch.ones
new = torch.arange(1, 6)                   # create whole numbers from 1 up to, but not including, 6
copy = torch.ones_like(new)                # create ones with the same shape and number type as new
print(copy)                                # display ones as whole numbers
print(torch.ones(5))                       # display ones created directly, as decimals
```

- Every tensor has a datatype, checked with `.dtype`
- `int` = whole numbers (counting); `float` = decimal numbers (measuring)
- A dot in the output (e.g. `1.`) means the tensor is a float
- The number in the name (e.g. 32, 64) is how many bits of memory each number uses
- Whole-number inputs default to `torch.int64`
- Decimal inputs, `zeros`, `ones` and `rand` default to `torch.float32`
- Neural networks mainly work in `torch.float32`
- A tensor has one dtype for all its numbers; one decimal makes the whole tensor float



- Set the datatype when creating a tensor with `dtype=`
- `dtype=` works with `torch.tensor`, `zeros`, `ones`, `rand`, `arange`, etc.
- `torch.float16`: 2 bytes per number, least exact, fastest, least memory
- `torch.float32`: 4 bytes per number, the default and the usual balance
- `torch.float64`: 8 bytes per number, most exact, slowest, most memory
- More bits = more precision but more memory and slower calculations
- float16 is used to speed up large models; float32 is the standard choice
No label + dots means float32.
No label + no dots means int64.
A dtype= label means it's something else, and the label tells you exactly what.

```python
# Q1: whole numbers stored as float32
arr = torch.tensor([1, 2, 3], dtype=torch.float32)             # create a tensor of whole numbers stored as 32-bit decimals
print(arr)                                                     # display the tensor

# Q2: zeros stored as float16
arr1 = torch.zeros(2, 3, dtype=torch.float16)                  # create a 2x3 tensor of zeros stored as 16-bit decimals
print(arr1.dtype)                                              # the datatype of the tensor

# Q3: compare precision of float16 and float64
float_16 = torch.tensor([2.71828182], dtype=torch.float16)     # store the number as a 16-bit decimal
print(float_16.item())                                         # display the stored value as a plain number
float_64 = torch.tensor([2.71828182], dtype=torch.float64)     # store the number as a 64-bit decimal
print(float_64.item())                                         # display the stored value as a plain number

# Q4: arange with and without a float dtype
arr3 = torch.arange(0, 5, dtype=torch.float32)                 # create numbers 0 to 4 as 32-bit decimals
print(arr3)                                                    # display the tensor
arr4 = torch.arange(0, 5)                                      # create numbers 0 to 4 as whole numbers (default)
print(arr4)                                                    # display the tensor
```

- `tensor.type(new_dtype)` creates a new tensor with the new datatype
- `tensor.to(new_dtype)` does the same thing
- The original tensor is unchanged unless you store the result back into it
- Converting float to int cuts off the decimal part; it does not round
- Mixing dtypes in `*` gives a result in the more precise type
- The 3 most common PyTorch errors: wrong datatype, wrong shape, wrong device
- Check them with `.dtype`, `.shape` and `.device`

```python
arr = torch.tensor([1.9,2.5,3.1])
print(arr.dtype)
print(arr)
new_arr = arr.type(torch.int64)
print(new_arr)

range =torch.arange(0,5)
print(range, range.dtype)
new_range = range.type(torch.float32)
print(new_range)

data1 = torch.tensor([1,2,3],dtype=torch.float16)
data2 = torch.tensor([3,4,5], dtype=torch.float32)
ans = data1 * data2
print(ans.dtype)

x = torch.tensor([1.5, 2.5])     # create a float32 tensor
x.type(torch.float16)            # create a float16 copy, but don't store it anywhere
print(x.dtype)                   # still torch.float32
```

- Attributes are information a tensor stores about itself
- `.dtype`: the datatype of the numbers
- `.shape` (or `.size()`): the size along each dimension
- `.device`: where the tensor is stored; `cpu` by default, `cuda` for a GPU
- These three match the 3 most common errors: wrong datatype, wrong shape, wrong device
- When something breaks, check `.dtype`, `.shape` and `.device` first
- f-strings (`f"text {value}"`) insert values into text for labelled printing
